```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tickets_sold = [92,180,165,220,128,450,75]
days_advertised = [28,35,21,42,30,18,14]
```

```python
data = sorted(zip(days_advertised, tickets_sold))

days_sorted, tickets_sorted = zip(*data)

plt.figure(figsize=(10,6))  

plt.plot(days_sorted, tickets_sorted, marker="o")

plt.xlabel("Days Advertised")
plt.ylabel("Tickets Sold")
plt.title("Tickets Sold vs Days Advertised")

plt.grid()

plt.show()
```
