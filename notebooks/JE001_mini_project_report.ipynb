```python
import pandas as pd
df = pd.read_csv("../data/JE001_real_sample.csv")
```

```python
df.head(8)
```

```python
print("How have ticket sales changed over time for Jigsaw Events Volume 001?")
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["date"], df["tickets_sold"])

plt.title("House Music Event Ticket Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Tickets Sold")

plt.xticks(rotation=45)
plt.grid()  

plt.show()
```

```python
plt.savefig("../outputs/JE001_ticket_sales_over_time.png")
```

```python
print("The graph shows that ticket sales generally increase over time, with sharper increases such as the one in early August. Such trend cannot be explained by the relationship between dates and ticket sales, but also factors such as significant marketing shifts, increased following, consumer confidence etc.  ")
```

```python
print("How does Instagram engagement relate to ticket sales for Jigsaw Events Volume 001?")
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(df["instagram_followers"], df["tickets_sold"])

plt.title("Instagram Engagement vs Ticket Sales")
plt.xlabel("Instagram Followers")
plt.ylabel("Tickets Sold")

plt.grid()
plt.show()
```

```python
plt.savefig("../outputs/JE001_ticket_sales_instagram_followers.png")
```

```python
print("The graph suggests that there is a positive, but weak correlation between Instagram followers and ticket sales. The graph shows a spread of points (no real correlation). However, generally as instagram followers increase, ticket sales also increase. This infers that as individuals see the following increasing, they may feel they will 'miss-out' and purchase a ticket as others follow the account and have purchased their ticket already. This is a common marketing tactic used to increase sales, and is known as 'social proof'.")
```
